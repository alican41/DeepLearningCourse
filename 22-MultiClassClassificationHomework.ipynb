```python
x = df[["Alcohol", "Malic acid", "Ash", "Alcalinity of ash", "Magnesium", "Total phenols", "Flavanoids", "Nonflavanoid phenols", "Proanthocyanins", "Color intensity", "Hue", "OD280/OD315 of diluted wines", "Proline "]].values
y = df["class"].values
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
y_test
```

```python
y_train
```

```python
x_train = torch.tensor(x_train, dtype=torch.float32)
x_test = torch.tensor(x_test, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)
```

```python
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
from torch import nn

class WineClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.layer_stack = nn.Sequential(
            nn.Linear(in_features=13, out_features=64),
            nn.ReLU(),
            nn.Linear(in_features=64, out_features=32),
            nn.ReLU(),
            nn.Linear(in_features=32, out_features=3)
        )

    def forward(self, x):
        return self.layer_stack(x)
    
    
```

```python
model = WineClassifier()
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
from torchmetrics.classification import MulticlassAccuracy
```

```python
accuracy = MulticlassAccuracy(num_classes=3)
```

```python
y_train = y_train - 1
y_test = y_test - 1
```

```python
epochs = 1000

for epoch in range(epochs):

    model.train()

    logits = model(x_train)
    loss = loss_fn(logits, y_train)

    pred = torch.softmax(logits, dim=1).argmax(dim=1)
    acc = accuracy(pred, y_train).item() * 100

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    model.eval()
    with torch.inference_mode():
        logits_test = model(x_test)
        loss_test = loss_fn(logits_test, y_test)

        pred_test = torch.softmax(logits_test, dim=1).argmax(dim=1)
        acc_test = accuracy(pred_test, y_test).item() * 100

    if epoch % 100 == 0:
        print(f"Epoch: {epoch} | Loss: {loss:.4f} | Acc: {acc:.2f}% | Test Loss: {loss_test:.4f} | Test Acc: {acc_test:.2f}%")
```
